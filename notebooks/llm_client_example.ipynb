## 🚀 Test 1: Automatische API-Auswahl

Der Client wählt automatisch die erste verfügbare API basierend auf vorhandenen Keys:
1. OpenAI (wenn `OPENAI_API_KEY` vorhanden)
2. Groq (wenn `GROQ_API_KEY` vorhanden)
3. Gemini (wenn `GEMINI_API_KEY` vorhanden)
4. Ollama (Fallback, nur lokal verfügbar)

```python
# Automatische API-Auswahl
client = LLMClient()

print(f"🤖 Verwendete API: {client.api_choice.upper()}")
print(f"📝 Modell: {client.llm}")
print(f"🌡️ Temperatur: {client.temperature}")
print(f"🔢 Max Tokens: {client.max_tokens}")
print()

# Einfache Test-Anfrage
messages = [
    {"role": "system", "content": "Du bist ein hilfreicher Assistent."},
    {"role": "user", "content": "Erkläre in einem Satz, was Machine Learning ist."},
]

print("💬 Sende Anfrage...")
response = client.chat_completion(messages)
print(f"\n📨 Antwort:\n{response}")
```

## 🔧 Test 2: OpenAI mit verschiedenen Modellen

Teste OpenAI mit verschiedenen GPT-Modellen:

```python
print("🧪 Test: OpenAI mit verschiedenen Modellen\n")
print("=" * 70)

openai_models = [
    "gpt-4o-mini",  # Schnell und günstig
    "gpt-4o",  # Leistungsstark
    "gpt-3.5-turbo",  # Klassiker
]

question = "Was ist Quantencomputing?"

for model in openai_models:
    print(f"\n🤖 Modell: {model}")
    print("-" * 70)

    try:
        client = LLMClient(api_choice="openai", llm=model, temperature=0.7, max_tokens=150)

        response = client.chat_completion([{"role": "user", "content": question}])

        print(f"✅ Antwort: {response}\n")

    except Exception as e:
        print(f"❌ Fehler: {e}\n")
```

## 🎛️ Test 5: Parameter-Variation

Teste wie verschiedene Parameter (Temperatur, max_tokens) die Ausgabe beeinflussen:

```python
print("🧪 Test: Einfluss verschiedener Parameter\n")
print("=" * 70)

# Wähle die erste verfügbare API
test_client = LLMClient(api_choice="groq")
api = test_client.api_choice
base_model = test_client.llm

print(f"Teste mit: {api.upper()} - {base_model}\n")

question = "Schreibe eine kurze Geschichte über einen Roboter."

# Test 1: Niedrige Temperatur (deterministisch)
print("\n🌡️ Test 1: Temperatur = 0.1 (deterministisch, fokussiert)")
print("-" * 70)
client_low_temp = LLMClient(api_choice=api, llm=base_model, temperature=0.1, max_tokens=150)
response1 = client_low_temp.chat_completion([{"role": "user", "content": question}])
print(response1)

# Test 2: Hohe Temperatur (kreativ)
print("\n\n🌡️ Test 2: Temperatur = 1.5 (kreativ, variabel)")
print("-" * 70)
client_high_temp = LLMClient(api_choice=api, llm=base_model, temperature=1.5, max_tokens=150)
response2 = client_high_temp.chat_completion([{"role": "user", "content": question}])
print(response2)

# Test 3: Kurze vs. lange Antworten
print("\n\n📏 Test 3: max_tokens = 50 (kurze Antwort)")
print("-" * 70)
client_short = LLMClient(api_choice=api, llm=base_model, temperature=0.7, max_tokens=50)
response3 = client_short.chat_completion([{"role": "user", "content": question}])
print(response3)

print("\n\n📏 Test 4: max_tokens = 500 (lange Antwort)")
print("-" * 70)
client_long = LLMClient(api_choice=api, llm=base_model, temperature=0.7, max_tokens=500)
response4 = client_long.chat_completion([{"role": "user", "content": question}])
print(response4)
```

## 💬 Test 6: Multi-Turn Konversation

Teste eine mehrteilige Konversation mit Kontext:

```python
print("🧪 Test: Multi-Turn Konversation\n")
print("=" * 70)

# Wähle Client
client = LLMClient(api_choice="groq", temperature=0.7, max_tokens=150)
print(f"Verwende: {client.api_choice.upper()} - {client.llm}\n")

# Konversations-Historie
conversation = [
    {"role": "system", "content": "Du bist ein Experte für künstliche Intelligenz."},
    {"role": "user", "content": "Was ist ein neuronales Netz?"},
]

# Turn 1
print("👤 User: Was ist ein neuronales Netz?")
response1 = client.chat_completion(conversation)
print(f"🤖 Assistant: {response1}\n")

# Füge Antwort zur Historie hinzu
conversation.append({"role": "assistant", "content": response1})

# Turn 2
conversation.append({"role": "user", "content": "Kannst du ein einfaches Beispiel geben?"})
print("\n" + "-" * 70)
print("👤 User: Kannst du ein einfaches Beispiel geben?")
response2 = client.chat_completion(conversation)
print(f"🤖 Assistant: {response2}\n")

# Füge Antwort zur Historie hinzu
conversation.append({"role": "assistant", "content": response2})

# Turn 3
conversation.append({"role": "user", "content": "Welche Frameworks gibt es dafür?"})
print("\n" + "-" * 70)
print("👤 User: Welche Frameworks gibt es dafür?")
response3 = client.chat_completion(conversation)
print(f"🤖 Assistant: {response3}")

print("\n" + "=" * 70)
print(f"✅ Konversation mit {len(conversation)} Nachrichten abgeschlossen")
```

## 📊 Test 7: API-Vergleich

Vergleiche die gleiche Anfrage über alle verfügbaren APIs:

```python
print("🧪 Test: API-Vergleich mit derselben Frage\n")
print("=" * 70)

question = "Erkläre Gradient Descent in 2-3 Sätzen."
print(f"❓ Frage: {question}\n")

apis = [
    ("openai", "gpt-4o-mini"),
    ("groq", "llama-3.1-8b-instant"),
    ("gemini", "gemini-3.1-flash-lite-preview"),
]

results = {}

for api_name, model in apis:
    print(f"\n🔄 Teste {api_name.upper()}...")
    print("-" * 70)

    try:
        client = LLMClient(api_choice=api_name, llm=model, temperature=0.7, max_tokens=200)

        response = client.chat_completion([{"role": "user", "content": question}])

        results[api_name] = response
        print(f"✅ {api_name.upper()}: {response}")

    except Exception as e:
        results[api_name] = None
        print(f"❌ {api_name.upper()}: Nicht verfügbar ({type(e).__name__})")

# Zusammenfassung
print("\n" + "=" * 70)
print("📋 Zusammenfassung:")
print("=" * 70)
successful = sum(1 for v in results.values() if v is not None)
print(f"✅ {successful}/{len(apis)} APIs erfolgreich getestet")

if successful == 0:
    print("\n⚠️ Keine API Keys gefunden!")
    print("Bitte füge API Keys als Secrets hinzu (🔑 Symbol im linken Menü)")
```

## 🎯 Test 8: System Prompts

Teste verschiedene System Prompts für unterschiedliches Verhalten:

```python
print("🧪 Test: Einfluss von System Prompts\n")
print("=" * 70)

client = LLMClient(api_choice="gemini", temperature=0.7, max_tokens=150)
print(f"Verwende: {client.api_choice.upper()} - {client.llm}\n")

question = "Was ist Python?"

system_prompts = [
    (
        "Technischer Experte",
        "Du bist ein technischer Experte und antwortest präzise und detailliert.",
    ),
    ("Freundlicher Lehrer", "Du bist ein geduldiger Lehrer, der komplexe Dinge einfach erklärt."),
    (
        "Kreativer Schriftsteller",
        "Du bist ein kreativer Schriftsteller und erzählst alles in Geschichtenform.",
    ),
]

for persona, system_prompt in system_prompts:
    print(f"\n🎭 Persona: {persona}")
    print("-" * 70)
    print(f'System Prompt: "{system_prompt}"\n')

    messages = [{"role": "system", "content": system_prompt}, {"role": "user", "content": question}]

    response = client.chat_completion(messages)
    print(f"💬 Antwort: {response}\n")
```
